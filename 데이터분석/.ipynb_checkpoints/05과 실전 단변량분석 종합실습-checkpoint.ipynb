    * 이상치 확인 및 조치 방안
    * 결측치 확인 및 조치 방안

* 비즈니스의 일반 사항과 특이사항을 도출
    * 추가 분석 대상 도출.

```python
path = 'https://raw.githubusercontent.com/JunhoYu/Data-Analysis/main/Carseats.csv'
data = pd.read_csv(path)
data.drop('Education', axis = 1, inplace = True)
data.head()
```

### ① Sales (Target) 

```python
var='Sales'
```

1) 변수의 비즈니스 의미

+ 각 지역 판매량
+ 1000개 단위

2) 숫자, 범주?

```python
data.info()
```

+ float형으로 숫자

3) NaN 존재 유무 및 조치방안

```python
data[var].isna().sum()
```

4) 가변수화 필요 여부

+ 숫자형 불필요

5) 기초통계량(수치화)

```python
data[var].describe()
```

6) 분포 확인(시각화)

```python
plt.figure(figsize=(50,4))
plt.subplot(1,3,1)
sns.countplot(data[var])

plt.figure(figsize=(20,4))
plt.subplot(1,3,2)
sns.kdeplot(data[var])

plt.figure(figsize=(20,4))
plt.subplot(1,3,3)
sns.boxplot(data[var])

plt.show()
```

+ 상위 7개의 sales가 있고, 그 값은 평균치보다 훨씬 높다 -> 인구수가 많은 지역?
