### 정적페이지 데이터 수집
네이버 연관 검색어 수집  
- bs4 : BeautifulSoup => HTML(css) -> css selector를 이용하여 엘리먼트 선택 

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

#### 1.웹서비스 분석 : URL

```python
keyword="kt"
url=f"https://search.naver.com/search.naver?&query={keyword}"
url
```

#### 2.request(url) -> response(html) -> HTML(str)

```python
response=requests.get(url)
response.text[:500]
```

#### 3.HTML(str) -> BeautifulSoup Object -> BS(CSS Selector) -> Data

```python
dom=BeautifulSoup(response.text,"html.parser")
type(dom)
```

```python
#select() : 엘리먼트 여러개 선택
#select_one() : 엘리먼트 한개 선택
elements=dom.select(".lst_related_srch>.item")
len(elements)
```
