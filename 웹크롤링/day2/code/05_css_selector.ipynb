### css selector
css 스타일을 적용시킬 HTML 엘리먼트를 선택하는 방법  


#### 1. Element Tag 이름으로 선택
css selector ->span -> python 3 선택

```python
%%html
<div class ="kt">
    <div>python 1</div>
    <p>python 2</p>
    <span>python 3</span>
</div>
```

#### 2. Tag의 id 값으로 선택
css selector -> #kt2 -> python2 선택

```python
%%html
<div class ="kt">
    <p id="kt1">python 1</p>
    <p  id="kt1">python 2</p>
    <p  id="kt1">python 3</p>
</div>
```

#### 3. Tag의 class값으로 선택
css selector -> .kt2 -> python2 선택  
css selector -> .kt -> python1,2,3 선택  
ctrl+enter 필수  


```python
%%html
<p class="kt kt1">python 1</p>
<p class="kt kt2">python 2</p>
<p class="kt kt3">python 3</p>
```

#### 4. Tag의 attr값으로 선택
css selectoer : [value="no1"]-> python1  

[value="no1"] {
    font-size: 20px;
}

```python
%%html
<p value="no1">python 1</p>
<p value="no2">python 2</p>
<p value="no3">python 3</p>
```

```python
# tag 이름 : span
# id : #kt
# class : .kt
# attr : [value="kt"]
```

### 여러개의 엘리먼트 선택
class no 엘리먼트를 모두 선택하고  
no2 클래스를 갖는 엘리먼트만 제외  
css selector : .no:not(.no2) => python 1, python 2

```python
%%html
<p class="no no1">python 1</p>
<p class="no no2">python 2</p>
<p class="no no3">python 3</p>
```

#### 5. n번째 엘리먼트 선택

css selector : .py:nth-child(2) -> python 2 선택  
 1부터 시작된다!!!  

<주의>  
.py 엘리먼트 중에서 2번째 : X  
2번째 엘리먼트 중에서 .py 클래스를 갖는 : O  

```python
%%html
<div>
    <p class="py">python 1</p>
    <span>python 0</span>
    <p class="py">python 2</p>
    <p class="py">python 3</p>
</div>
```

#### 6. 계층적으로 엘리먼트 선택
.wrap >p : inner 1 선택  
.wrap p :inner 1, inner 2 선택 -> 모든 p 선택

```python
%%html
<div class ="wrap">
    <p>inner 1</p>
    <div>
        <p>inner 2</p>
    </div>
</div>
</div>
```

#### 여러개의 엘리먼트 선택
.py1,.py3 : python1, python3 선택 -> 콤마 앞뒤의 모든 엘리먼트 선택

```python

%%html
<p class="py1">python 1</p>
<p class="py2">python 2</p>
<p class="py3">python 3</p>
```

### summary
엘리먼트 하나 선택
+ tag 이름 : span
+ id : #kt
+ class : .kt
+ attr : [value="kt"]

여러개의 엘리먼트 선택
+ :not(.kt1)
+ :nth-child(n)
+ .wrap >p : 한단계 하위 엘리먼트
+ .wrap p : 모든 하위 엘리먼트
+ .kt1,.kt2 : 선택하는 모든 엘리먼트 선택
