### gmarket
베스트 상품 200개 수집

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
#1. URL
# 새로고침 -> 맨 위
url="http://corners.gmarket.co.kr/Bestsellers"
```

```python
#2. request -> response : HTML
response=requests.get(url)
response
```

```python
#3. html(str) -> beautifulsoup object -> css-selector -> df
dom=BeautifulSoup(response.text,"html.parser")
type(dom)
```

```python
#200개 상품 엘리먼트 선택
#크롬 항목 선택 -> 오른쪽 클릭 -> copy -> copy selector
elements=dom.select("#gBestWrap > div > div:nth-child(5) > div > ul >li")
len(elements)
```

```python
element=elements[0]
```

```python
element
```

```python
data={
    "title":element.select_one(".itemname").text,
    "o_price":element.select_one(".o-price").text,
    "s_price":element.select_one(".s-price>strong").text,
    "img":"http:"+element.select_one("img").get("data-original"),
}
data
```

```python
items=[]
for element in elements:
    data={
        "title":element.select_one(".itemname").text,
        "o_price":element.select_one(".o-price").text,
        "s_price":element.select_one(".s-price>strong").text,
        "img":"http:"+element.select_one("img").get("data-original"),
    }
    items.append(data)
items[:2]
    
```

```python
df=pd.DataFrame(items)
df.tail(2)
```
