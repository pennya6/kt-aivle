### API 
- application programming interface
- api를 사용해서 데이터를 수집하는 것은 서비스에 데이터를 제공하는 공식적인 방법으로 데이터 수집
- naver api : papago api -> 번역 서비스

1. APP 등록 -> app_key(request_token)
2. Naver API Document check -> URL
3. request(url, app_key) -> response(json) -> json(str)
4. json(str) -> list, dict -> dataframe

```python
import pandas as pd
import requests,json
```

#### ID,PW

```python
CLIENT_ID, CLIENT_SECRET = "za1kdvINTuafBgNr_uM_", "yJfgi3NSn6"
```

#### url

```python
url="https://openapi.naver.com/v1/papago/n2mt"
```

#### params, txt, head

```python
txt="파이썬은 재미있습니다."
params={
    "source":"ko",
    "target":"en",
    "text":txt
}
header={
    "Content-Type":"application/json",
    "X-Naver-Client-Id":CLIENT_ID,
    "X-Naver-Client-Secret":CLIENT_SECRET,
}
```

#### response
##### dumps()

```python
#json.dumps() : 인터넷 트래픽에서는 영문, 숫자, 특수문자만 사용가능
# 한글과 같은 문자를 인코딩(영문, 숫자, 특수문자)
response=requests.post(url,json.dumps(params),headers=header)
```

#### 함수

```python
def translate(txt):
    CLIENT_ID, CLIENT_SECRET = "za1kdvINTuafBgNr_uM_", "yJfgi3NSn6"
    params={"source":"ko","target":"en","text":txt}
    header={ "Content-Type":"application/json","X-Naver-Client-Id":CLIENT_ID,
            "X-Naver-Client-Secret":CLIENT_SECRET}
    response=requests.post(url,json.dumps(params),headers=header)
    txt_en=response.json()["message"]["result"]["translatedText"]
    return txt_en
```

```python
translate("정말 힘들다")
```
