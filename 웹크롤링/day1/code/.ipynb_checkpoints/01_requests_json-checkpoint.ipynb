### webcrawling
- 웹 크롤링의 종류
    - 정적페이지 : 페이지의 데이터가 변경될때 url이 변경 , html 포맷
    - 동적페이지 : 페이지의 데이터가 변경될때 url이 변경되지 않음 , json 포맷
- request package
    - 브라우저의 URL을 입력하면 서버에서 데이터를 다운받아 화면에 출력 : URL -> Data
    - request package : URL -> Data  즉 브라우저 기능을 파이썬으로 구현


### Naver Stock Data
- Kospi 지수
- Kosdaq 지수
- USD : 원달러 환율

```python
import requests
import pandas as pd
```

 1. 웹서비스를 분석 : 크롬 개발자 도구 -> URL 최종 획득가능
 2. Requests(데이터요청) -> response (데이터 얻어오기) -> JSON(str)
 3. JSON(str) ->list, dict -> Dataframe

```python
# 1. 웹서비스를 분석 : 크롬 개발자 도구 -> URL 최종 획득가능
# 크롬개발자 도구 -> clear -> 더보기 버튼 -> head의 url, post등 확인
url = "https://m.stock.naver.com/api/index/KOSPI/price?pageSize=10&page=4"
```

```python
# 2. requests(데이터요청) -> response (데이터 얻어오기) -> JSON(str)
response=requests.get(url)
response
```

```python
response.text[:200]
```

```python
# 3. JSON(str) ->list, dict -> Dataframe
data=response.json()
type(data),data[:1]
```

```python
#데이터프레임
df=pd.DataFrame(data)[["localTradedAt","closePrice"]]
df
```

#### 함수 만들기

```python
#4. 함수 만들기 : params = pagesize,page
def stock_price(pagesize,page,code="KOSPI"):
    """This function is crawling stock price from webpage.
    Param
    ------
    pagesize : int , one page size
    page : int, page number
    code : string, kospi or kosdaq
    
    Return
    ------
    type : Dataframe, display date, price columns
    """
    url=f"https://m.stock.naver.com/api/index/{code}/price?pageSize={pagesize}&page={page}"
    response=requests.get(url)
    data=response.json()
    return pd.DataFrame(data)[["localTradedAt","closePrice"]]
```

```python
df=stock_price(30,2)
```

```python
df
```

#### 원달러 환율 데이터 수집
증권 -> 시장지표 -> 미국 USD -> 일별시세(더보기) -> 개발자 도구  
최근 60일치 원달러 환율 데이터 수집

```python
def usd(page,pagesize):
    url3=f"https://api.stock.naver.com/marketindex/exchange/FX_USDKRW/prices?page={page}&pageSize={pagesize}"
    response3=requests.get(url3)
    usd=response3.json()
    usd=pd.DataFrame(usd)[["localTradedAt","closePrice"]]
    return usd
```

```python
usd=usd(1,60)
```

### 간단한 데이터 분석
상관관계분석 : 두 데이터 집합 사이에 어떤 관계가 있는 확인하는 분석 방법
ex) 원달러 환율이 높으면 코스피, 코스닥 주가 지수가 낮다. -> 음의 상관관례를 갖는다.

#### 피어슨 상관계수 : df.corr()
-> 결과가 1과 가까울수록 강한 양의 상관관계를 갖는다.
-1 과 가까울수록 강한 음의 상관계수를 갖는다.
0과 가까울수록 관계가 없다.


```python
# 데이터 전처리
df=kospi.copy()
df["kosdaq"]=kosdaq["closePrice"]
df["usd"]=usd["closePrice"]
df=df.rename(columns={"closePrice":"kospi"})
df.tail(2)
```

```python
df.dtypes
```

```python
#데이터 타입 float로 변경
#df[column].apply() : 모든 데이터를 함수에 대입한 결과를 출력
df['kospi']=df['kospi'].apply(lambda data: float(data.replace(",",""))) # , 제거후 형 변환
df['usd']=df['usd'].apply(lambda data: float(data.replace(",","")))
df['kosdaq']=df['kosdaq'].apply(lambda data: float(data.replace(",","")))
```

```python
# kospi - kosdaq : 0.984 -> 1과 가까우며 강한 양의 상관관계
# kospi - usd : -0.878 : -1과 가까우며 강한 음의 상관관계
df[['kospi', 'kosdaq', 'usd']].corr()
```

### pandas add

#### copy()

```python
# copy() -> copy를 사용해야 원본 수정 에러 X
d1=[1,2,3]
d2=d1
print(d1,d2) #[1, 2, 3] [1, 2, 3]
d1[1]=4
print(d1,d2) #[1, 4, 3] [1, 4, 3]
```

```python
d1=[1,2,3]
d2=d1 #주소값 복사 -> 얕은 복사 (call by reference)
d3=d1.copy() #값복사 -> 깊은 복사 (call by value)
print(d1,d2,d3) #[1, 2, 3] [1, 2, 3] [1, 2, 3]
d1[1]=4
print(d1,d2,d3) #[1, 4, 3] [1, 4, 3] [1, 2, 3]
```

#### apply(func)
모든 데이터를 func을 적용시킨 결과 출력

```python
#apply(func) : 모든 데이터를 func을 적용시킨 결과 출력
df=pd.DataFrame([
    {"age":23},{"age":36},{"age":27}
])
df
```

```python
# 연령대 컬럼 추가
def change_ages(age):
    return age // 10*10 
```

```python
df["ages"]=df["age"].apply(change_ages)
df
```

#### lambda : 일회성함수
사용하는 이유 -> 간단한 함수(파라미터를 받아서 바로 리턴하는)를 메모리를 절약하여 사용가능

```python
#함수 3개 사용, 메모리 3칸 사용
def plus(n1,n2):
    return n1+n2
def minus(n1,n2):
    return n1-n2
def calc(func,n1,n2):
    return func(n1,n2)
```

```python
calc(plus,1,2),calc(minus,1,2)
```

```python
plus_lambda=lambda n1,n2 : n1+n2
plus(2,3), plus_lambda(2,3)
```

```python
def calc(func,n1,n2):
    return func(n1,n2)
calc(lambda n1,n2: n1+n2,1,2), calc(lambda n1,n2: n1-n2,1,2)
```

### Summary
웹페이지 종류
1. 정적페이지 : 데이터가 변경될떄 URL 변경 O -> HTML
2. 동적페이지 : 데이터가 변경될때 URL 변경 X -> JSON

동적 웹크롤링 절차
1. 웹서비스 분석 : 크롬 개발자 도구 -> URL
2. request(url) -> response(json) -> json(str)
3. json(str) -> list, dict -> dataframe
