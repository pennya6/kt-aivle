```python
# 클래스로 이터레이터 만들기 : 피보나치 수열
```

```python
class Fib:
    def __init__(self):
        self.prev=0
        self.curr=1
    def __iter__(self):
        return self
    def __next__(self):
        value=self.curr
        self.curr+=self.prev
        self.prev=value
        return value
```

```python
fib=Fib()
type(fib)
```

```python
next(fib)
```

```python
# yield : 일반 함수를 제너레이터로 만들어 주는 명령어
# next 함수를 실행해서 yield를 만나면 코드 실행 중단
```

```python
def fib():
    prev,curr=0,1
    while True:
        yield curr #curr를 return하고 코드 실행 중단
        prev, curr=curr,prev+curr
```

```python
0 1 1 2 3 5 8 13
```

```python
f=fib()
```

```python
type(f)
```

```python
next(f)
```
