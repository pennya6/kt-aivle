#### 브라우저 자동화

```python
driver=webdriver.Chrome("chromedriver.exe")
driver.get("https://daum.net")
driver.set_window_size(200,600)
driver.execute_script("window.scrollTo(200,300);")
driver.execute_script("alert('hello selenium!');")
alert=driver.switch_to.alert
alert.accept() #페이지를 파이썬코드로 !! 컨트롤해야햄
driver.execute_script("window.scrollTo(0,0);")
#새로운 다음창을 만들어서 css selector 확인
#q -> selector
driver.find_element(By.CSS_SELECTOR,"#q").send_keys("python")
driver.find_element(By.CSS_SELECTOR, ".inner_search > .ico_pctop.btn_search").click()
driver.quit()
```

### 크롤링 방법 정리
- 1. <json, requests> 웹페이지의 API 트래픽 분석 -> 데이터 수집 : naver stock
- 2. <json, requests> 공식적으로 제공하는 API -> application key -> 데이터 수집 : naver api papago, trend
- 3. <html, BeatufulSoup(css selector) requests> 웹페이지의 html 코드 받아서 데이터 수집 : gmarket
- 4. <selenium : browser - python> 브라우저를 파이썬 코드로 컨트롤해서 데이터 수집 : ted
- <mark>좋은 순서  : 2 > 1 > 3 > 4</mark>

### 간단한 머신러닝

```python
df=pd.read_csv("data/premierleague.csv")
df.head(2)
```

```python
feature=df[['gf','ga']]
target=df['points']
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression().fit(feature,target)
```

```python
#데이터 예측 : 득점 80, 실점 36
np.round(model.predict([[80,36]]))
```

```python
#warning 메세지 지우기
import warnings
warnings.filterwarnings("ignore")
```

```python
# data > train(8h) > model(ram) > computer off > train(8h)
# model > ssd(hdd)
import pickle
```

```python
# ram > ssd(hdd)
with open("model.pkl","wb")as file:
    pickle.dump(model,file)
```

```python
# ssd > ram
with open("model.pkl","rb") as file:
    load_model=pickle.load(file)
```

```python
np.round(load_model.predict([[80,30]]))
```

### 텍스트 데이터 수집
ted 사이트 : https://ted.com/talks

#### 1.브라우저 열기

```python
driver=webdriver.Chrome()
```

#### 2. 페이지 이동

```python
driver.get("https://ted.com/talks")
```

#### 3. 제목 데이터 텍스트 수집

```python
sub_title=driver.find_element(By.CSS_SELECTOR,"#banner-secondary").text
```

```python
sub_title
```

#### 4.select box에서 한국어 메뉴 클릭

```python
# languages>optgroup>[lang="ko"]
driver.find_element(By.CSS_SELECTOR,'#languages [lang="ko"]').click()
```

#### 5.전체 데이터의 제목과 링크 데이터 수집

```python
elements=driver.find_elements(By.CSS_SELECTOR,"#browse-results>.row>div")
len(elements)
```

```python
element=elements[0]
title=element.find_element(By.CSS_SELECTOR,"h4>.ga-link").text
link=element.find_element(By.CSS_SELECTOR,"h4>.ga-link").get_attribute("href")
title ,link
```

#### 6.데이터프레임으로 만들기

```python
data=[]
for element in elements:
    data.append({
        "title":element.find_element(By.CSS_SELECTOR,"h4>.ga-link").text,
        "link":element.find_element(By.CSS_SELECTOR,"h4>.ga-link").get_attribute("href")
    })
df=pd.DataFrame(data)
df.tail(2)
```

```python
driver.quit()
```

### Headless
- 브라우저를 화면에 띄우지 않고 메모리 상에서만 브라우저를 실행하여 크롤링하는 방법
- window가 지원되지 않는 환경에서 selenium 사용가능
- 크롬 버전 : 60 이상부터 지원

#### 크롬 버전 확인

```python
driver=webdriver.Chrome()
version=driver.capabilities["browserVersion"]
driver.quit()
version
```

```python
options=webdriver.ChromeOptions()
options.add_argument("headless")

driver=webdriver.Chrome(options=options)

driver.get("https://ted.com/talks")
sub_title=driver.find_element(By.CSS_SELECTOR,"#banner-secondary").text
driver.quit()
sub_title
```
