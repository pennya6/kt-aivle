# 1. 데이터 전처리

#### [실습문제1] 열 추가 
* 기존 '기준일ID'열에서 날짜 데이터를 추출하여 새로운 열로 추가 합니다.
    - pd.to_datetime() 함수를 사용해 '기준일ID'열을 datetime 형식으로 변경하겠습니다.
    - year, month, day 열로 기존 데이터 프레임에 연월일 정보를 저장합니다.
    - 데이터 프레임 열에 대해 날짜 정보를 얻을 때에는 **dt.year**, **dt.month**, **dt.day** 속성을 사용합니다. 

```python
# 아래에 실습코드를 작성하고 결과를 확인합니다.
df_total['기준일ID']=pd.to_datetime(df_total['기준일ID'].astype('str'))
```

```python
df_total.head()
```

```python
from datetime import date
df_total["년"] = [d.year for d in df_total["기준일ID"]]
df_total["월"] = [d.month for d in df_total["기준일ID"]]
df_total["일"] = [d.day for d in df_total["기준일ID"]]
```

```python
df_test['기준일ID']=pd.to_datetime(df_test['기준일ID'].astype('str'))
df_test["년"] = [d.year for d in df_test["기준일ID"]]
df_test["월"] = [d.month for d in df_test["기준일ID"]]
df_test["일"] = [d.day for d in df_test["기준일ID"]]
```

```python
df_test.tail()
```

#### [실습문제2] 불필요한 열 제거 및 열 이름 변경
* df_total과 df_test에서 '총생활인구수'외에 불필요한 열을 제거 하도록 하겠습니다.
* 시간 데이터를 남기고, '시간대구분' column 을 'hour'로 rename 하도록 하겠습니다. 

```python
# 아래에 실습코드를 작성하고 결과를 확인합니다.
df_total.rename(columns={'시간대구분':'hour'},inplace=True)
df_test.rename(columns={'시간대구분':'hour'},inplace=True)
```

#### [실습문제3] 시프트 & 롤링
* shift()를 활용하여 이동할 행 수를 지정합니다.
* rolling()을 활용하여 추가할 이동평균 수를 지정합니다.

```python
# 아래에 실습코드를 작성하고 결과를 확인합니다.
# 1416 -> 이전 데이터로 1416시간 치의 예측치를 보겠다
df_total['2m'] = df_total['총생활인구수'].shift(1416)
df_total['1m'] = df_total['총생활인구수'].shift(708)
df_total['1d'] = df_total['총생활인구수'].shift(24)
df_total['1h'] = df_total['총생활인구수'].shift(1)
#df_total['2d_y'] = df_total['총생활인구수'].shift(24*7)
```

```python
# 아래에 실습코드를 작성하고 결과를 확인합니다.
df_test['2m'] = df_test['총생활인구수'].shift(1416)
df_test['1m'] = df_test['총생활인구수'].shift(708)
df_test['1d'] = df_test['총생활인구수'].shift(24)
df_test['1h'] = df_test['총생활인구수'].shift(1)
```

```python
df_total['2m_mean']=df_total['총생활인구수'].rolling(window=1416).mean()
df_test['2m_mean']=df_test['총생활인구수'].rolling(window=1416).mean()
```

#### [실습문제4] 결측치 처리 및 불필요한 열 제거
* 롤링 & 시프트 과정에서 발생한 결측치를 처리하도록 하겠습니다.
* 불필요한 열을 추가로 제거 하겠습니다.(시간 관련 행들)

```python
# 아래에 실습코드를 작성하고 결과를 확인합니다.
# df_total.dropna(axis=0,inplace=True)
# df_test.dropna(axis=0,inplace=True)
df_total.fillna(df_total.mean(),inplace=True)
df_test.fillna(df_test.mean(),inplace=True)
```

```python
# 아래에 실습코드를 작성하고 결과를 확인합니다.
df_total.drop(labels='기준일ID',axis=1,inplace=True)
df_test.drop(labels='기준일ID',axis=1,inplace=True)
```

#### [실습문제5] train, test 데이터 나누기
* df_total과 df_test를 각각 train_x, train_y, test_x, test_y 로 나누어 주세요.
* 인덱스를 리셋 해주세요.

```python
df=pd.concat([df_total,df_test])
```

```python
df.info()
```

```python
#x=df.drop(['y'],axis=1)
x=df[['hour','년','월','일','2m','1d','2m_mean','1h','1m']]
y=df.loc[:,'총생활인구수']
```

```python
x
```

```python
# 아래에 실습코드를 작성하고 결과를 확인합니다.
from sklearn.model_selection import train_test_split
train_x, test_x, train_y, test_y = train_test_split(x, y, test_size = df_test.shape[0]-1, shuffle = False) #30 : 자연수 - 개수, shuffle : 섞어라 
```
