6) 반복문을 이용하여 100이하의 소수(prime number)를 리스트에 담아 봅시다.
* 소수(prime number) : 약수가 오직 1과 자기자신 만 존재하는 수. 

```python
#에라토스테네스의 체 
#지금 방법은 단순
prime_number=[]
for i in range(2,101):
    divied=0 #약수의 몇개인가
    for j in range(1, i+1) :
        if i%j==0:
            divied+=1
    if divied==2:
        prime_number.append(i)
```

```python
prime_number
```
