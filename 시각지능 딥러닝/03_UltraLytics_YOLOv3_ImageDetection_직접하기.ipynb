## Image Detection

3. detect.py를 python으로 직접 호출하여 수행

### 1) 원하는 이미지 다운로드

```python
!wget -O /content/yolov3/pretrained/yolov3-spp.pt https://github.com/ultralytics/yolov3/releases/download/v9.6.0/yolov3-tiny.pt
```

### 3) detect.py 실행

- 명령어 도움말 : python detect.py -h

```python
!cd yolov3; python detect.py -h
```

```python
!wget -O /content/yolov3/data/images/14th_street.jpg https://raw.githubusercontent.com/DrKAI/image/main/14th_Street_2005.jpg
```

```python
!cd yolov3; python detect.py \
    --weights '/content/yolov3/pretrained/yolov3-spp.pt' \
    --source '/content/test2.PNG' \
    --project '/content/yolov3/detected' \
    --name 'images' \
    --img 640 \
    --conf-thres 0.5 \
    --iou-thres 0.4 \
    --line-thickness 2 \
    --exist-ok \
    --device CPU
```
