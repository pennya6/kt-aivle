```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
x = cancer.data
y = cancer.target
```

```python
x.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=2022,test_size=0.2)
```

```python
#####################
# 라이브러리 불러오기
#####################
import tensorflow as tf
import numpy as np
from tensorflow import keras
```

```python
########
# 모델링
########
keras.backend.clear_session()
il=keras.layers.Input(shape=(30,),name='inputLayer')
ol=keras.layers.Dense(1,activation='sigmoid')(il)
model=keras.models.Model(inputs=il,outputs=ol)
model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics=['accuracy'])
```

```python
model.summary()
```

```python
###########
# 모델 학습
###########
model.fit(x_train,y_train,epochs=10)
```

```python
###########
# 모델 예측
###########
y_pred=model.predict(x_test)
```

# 히든 레이어 추가하여 모델링!

    히든 레이어 2개를 추가하여 다시 모델링!
    히든 레이어 노드 수 : 32
    히든 레이어 activation = 'relu'

```python
########
# 모델링
########
keras.backend.clear_session()
il=keras.layers.Input(shape=(30,))
hidden=keras.layers.Dense(32,activation='relu')(il)
hidden2=keras.layers.Dense(32,activation='relu')(hidden)
ol=keras.layers.Dense(1,activation='sigmoid')(hidden2)
model=keras.models.Model(inputs=il,outputs=ol)
model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics=['accuracy'])
```

```python
model.summary()
```

```python
###########
# 모델 학습
###########
model.fit(x_train,y_train,epochs=10)
```

```python
###########
# 모델 예측
###########
y_pred=model.predict(x_test)
```

```python
y_pred= np.where(y_pred >= 0.5, 1, 0)
```
