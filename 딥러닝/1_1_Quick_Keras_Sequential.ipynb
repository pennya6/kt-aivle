## Quick Linear Regression!

회사내에서 바로 사용하고 싶다면 tensorflow  

```
import tensorflow as tf
from tensorflow import keras

import numpy as np
```

```
x = np.array(range(0,20))
y = x * 2 -1 

print(x)
print(y)
```

```
x.shape, y.shape #숫자 1이 있다고 생각해라 
```

((20,)) -> 1 생략  
input  
(shape(1,))  
output  
Dense(1)

모델링 4가지 단계

```
# 1번 청소 : 이미 만들어진 모델이 있다면 그 모델을 없애줘
keras.backend.clear_session()

# 2번 모델 선언, 레이어 블록을 쌓을 발판 생성
model = keras.models.Sequential()

# 3번 모델 블록 조립
model.add( keras.layers.Input(shape=(1,)) ) # input layer
model.add( keras.layers.Dense(1) ) #output layer

## 오리지널 Sequential API
# model.add( keras.layers.Dense(1, input_shape=(1,)) )

# 4번 컴파일 : 모델이 어떻게 동작할지 알려주는 단계
# loss = cost 평가 지표, 무엇으로 비교할지를 정하는 것
# 회귀 -> mse 평가 지표 문자열 형태로 넣기 
# 옵티마이저 = 경사하강법 , 얼마나 이쁘게 진행할지를 정하는 단계
model.compile(loss='mse', 
              optimizer='adam')
```

```
# 모델학습
# 학습 과정이기 때문에 x,y 
# epochs : 학습 데이터를 몇번이나 학습 시킬래? 
# verbose : 1 -> 학습 과정 processbar로 보여줌 
model.fit(x, y, epochs=10, verbose=1)
```

```
print(y)
```

```
# 모델 예측 
model.predict(x).reshape(-1)
```

## Quick Logistic Regression!

```
import tensorflow as tf
import tensorflow.keras as keras

import numpy as np
```

```
x = np.array(range(0,20)) 
y = np.array([0]*10 + [1]*10)
print(x)
print(y)
```

```
# 혹시 이미 그려둔 그래프가 있다면 날려줘!
keras.backend.clear_session()

# model에 순차적으로 레이어를 쌓아가겠다는 의도!
model = keras.models.Sequential()

# model에 인풋 값을 받는 레이어를 넣음
model.add( keras.layers.Input(shape=(1,)) )
# model에 Dense 레이어를 넣을거야 (최초의 레이어) : weight를 곱하고, bias를 더해주는 과정
model.add( keras.layers.Dense(1, activation='sigmoid') )


# 오리지널 Sequential API
# model.add( keras.layers.Dense(1, input_shape=(1,), activation='sigmoid') )

# 컴파일 해주렴
# 이진 분류 -> crossentropy
model.compile(loss = 'binary_crossentropy',
              optimizer = 'adam',
              metrics=['accuracy'])
```

```
# 데이터를 넣어서 학습시키자!
#model.fit(x[:15], y[:15], epochs=10, verbose=1)
model.fit(x, y, epochs=10, verbose=1)
# 결과 출력해줘!
print(y[:5])
print(model.predict(x[:5]).reshape(-1)) # threshold가 0.5보다 크면 1이 된당 하지만 여기서는 확률값으로 보여줌 
```

0과 1이 아닌 확률값으로 보여줌  
loss는 계속해서 줄어들어야 정상

```
keras.backend.clear_session()
```

```
from numpy.core.fromnumeric import shape
model=keras.models.Sequential()
model.add( keras.layers.Input(shape=(1,)) ) #input이 functional이기 때문
model.add( keras.layers.Dense(1, activation='sigmoid') )
```

```
model.summary()
```
